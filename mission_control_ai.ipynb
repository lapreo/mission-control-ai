```python
import ollama
import random
import time
from datetime import datetime

LIMITES = {
    "temperatura_max": 85.0,
    "energia_min": 20.0,
    "sinal_min": 30.0,
    "pressao_min": 95.0,
    "oxigenio_min": 18.0,
}
NOME_MISSAO = "ARES-7"

print("✅ Configurações carregadas com sucesso.")
print(f"🚀 Missão: {NOME_MISSAO}")
```

## 🔢 Célula 3 — Geração de Dados Simulados da Missão

```python
def gerar_dados_missao(cenario="normal"):
    """
    Gera dados simulados da missão espacial.
    Cenários disponíveis: 'normal', 'critico', 'aleatorio'
    """
    if cenario == "normal":
        dados = {
            "temperatura":  round(random.uniform(20.0, 70.0), 1),
            "energia":      round(random.uniform(40.0, 95.0), 1),
            "sinal":        round(random.uniform(60.0, 99.0), 1),
            "pressao":      round(random.uniform(98.0, 105.0), 1),
            "oxigenio":     round(random.uniform(20.0, 23.0), 1),
            "status_modulo": random.choice(["OPERACIONAL", "OPERACIONAL", "OPERACIONAL", "STANDBY"]),
        }
    elif cenario == "critico":
        dados = {
            "temperatura":round(random.uniform(86.0, 120.0), 1),
            "energia":round(random.uniform(5.0, 19.9), 1),
            "sinal":round(random.uniform(5.0, 29.9), 1),
            "pressao":round(random.uniform(70.0, 94.9), 1),
            "oxigenio":round(random.uniform(14.0, 17.9), 1),
            "status_modulo":random.choice(["FALHA", "FALHA", "EMERGÊNCIA"]),
        }
    else:
        dados = {
            "temperatura":round(random.uniform(15.0, 125.0), 1),
            "energia":round(random.uniform(3.0, 100.0), 1),
            "sinal":round(random.uniform(5.0, 100.0), 1),
            "pressao":round(random.uniform(70.0, 110.0), 1),
            "oxigenio":round(random.uniform(13.0, 24.0), 1),
            "status_modulo": random.choice(["OPERACIONAL", "STANDBY", "FALHA", "EMERGÊNCIA"]),
        }

    dados["timestamp"] = datetime.now().strftime("%Y-%m-%d %H:%M:%S")
    return dados

print("Função de geração de dados pronta.")
```

## 🚨 Célula 4 — Lógica de Alertas e Tomada de Decisão

```python
def verificar_alertas(dados):
    """
    Verifica parâmetros críticos e retorna lista de alertas.
    Implementa lógica de tomada de decisão automatizada.
    """
    alertas = []
    acoes = []

    if dados["temperatura"] > LIMITES["temperatura_max"]:
        alertas.append(f"🌡️  TEMP CRÍTICA: {dados['temperatura']}°C (máx: {LIMITES['temperatura_max']}°C)")
        acoes.append("→ Acionando sistema de resfriamento de emergência")
        acoes.append("→ Reduzindo carga computacional em 40%")

    if dados["energia"] < LIMITES["energia_min"]:
        alertas.append(f"⚡ ENERGIA BAIXA: {dados['energia']}% (mín: {LIMITES['energia_min']}%)")
        acoes.append("→ Ativando MODO DE ECONOMIA DE ENERGIA")
        acoes.append("→ Desligando sistemas não essenciais")
        acoes.append("→ Redirecionando painéis solares para carga máxima")

    if dados["sinal"] < LIMITES["sinal_min"]:
        alertas.append(f"📡 SINAL FRACO: {dados['sinal']}% (mín: {LIMITES['sinal_min']}%)")
        acoes.append("→ Reorientando antena de comunicação")
        acoes.append("→ Ativando protocolo de comunicação de backup")

    if dados["pressao"] < LIMITES["pressao_min"]:
        alertas.append(f"🔧 PRESSÃO BAIXA: {dados['pressao']} kPa (mín: {LIMITES['pressao_min']} kPa)")
        acoes.append("→ ALERTA DE VEDAÇÃO — verificar integridade do módulo")
        acoes.append("→ Ativando compartimento pressurizado de contingência")

    if dados["oxigenio"] < LIMITES["oxigenio_min"]:
        alertas.append(f"💨 OXIGÊNIO CRÍTICO: {dados['oxigenio']}% (mín: {LIMITES['oxigenio_min']}%)")
        acoes.append("→ PROTOCOLO DE EMERGÊNCIA DE SUPORTE DE VIDA ATIVADO")
        acoes.append("→ Acionando reservatório de O₂ de emergência")

    if dados["status_modulo"] in ["FALHA", "EMERGÊNCIA"]:
        alertas.append(f"⚠️  MÓDULO: {dados['status_modulo']}")
        acoes.append("→ Executando diagnóstico completo do módulo")
        acoes.append("→ Notificando equipe de engenharia na Terra")

    return alertas, acoes


def nivel_criticidade(alertas):
    """Determina o nível de criticidade geral da missão."""
    n = len(alertas)
    if n == 0:
        return "✅ NORMAL", "verde"
    elif n <= 2:
        return "⚠️  ATENÇÃO", "amarelo"
    elif n <= 4:
        return "🚨 CRÍTICO", "laranja"
    else:
        return "🆘 EMERGÊNCIA", "vermelho"

print("Sistema de alertas e tomada de decisão pronto.")
```

## 🤖 Célula 5 — Integração com IA (Llama via Ollama)

```python
def analisar_com_ia(dados, alertas):
    """
    Envia os dados da missão para o modelo Llama e obtém análise inteligente.
    """
    prompt_usuario = f"""Dados atuais da missão {NOME_MISSAO}:
- Temperatura: {dados['temperatura']}°C
- Energia (bateria): {dados['energia']}%
- Sinal de comunicação: {dados['sinal']}%
- Pressão interna: {dados['pressao']} kPa
- Nível de oxigênio: {dados['oxigenio']}%
- Status do módulo: {dados['status_modulo']}
- Alertas detectados: {len(alertas)}
{'- ALERTAS: ' + ' | '.join(alertas) if alertas else '- Sem alertas críticos'}

Faça uma análise situacional em 3-4 frases. Seja direto e técnico."""

    try:
        resposta = ollama.chat(
            model="llama3.2:1b",
            messages=[
                {
                    "role": "system",
                    "content": (
                        f"Você é o sistema de IA do Mission Control da missão espacial {NOME_MISSAO}. "
                        "Analise dados operacionais e forneça avaliações técnicas precisas. "
                        "Seja objetivo, use terminologia aeroespacial e indique prioridade de ações. "
                        "Responda sempre em português."
                    )
                },
                {
                    "role": "user",
                    "content": prompt_usuario
                }
            ]
        )
        return resposta["message"]["content"]
    except Exception as e:
        return f"[ERRO NA ANÁLISE DA IA: {str(e)}]"

print("Integração com IA (Llama) configurada.")
```

## 🖥️ Célula 6 — Display do Status da Missão

```python
def exibir_status(dados, alertas, acoes, analise_ia):
    """Exibe o painel completo de status da missão."""
    nivel, cor = nivel_criticidade(alertas)
    linha = "═" * 55
    linha2 = "─" * 55

    print(f"\n{linha}")
    print(f"  🚀  MISSION CONTROL AI — {NOME_MISSAO}")
    print(f"  📅  {dados['timestamp']}")
    print(linha)

    print("\n  📊  PARÂMETROS OPERACIONAIS")
    print(linha2)

    def barra(valor, maximo=100, tamanho=20):
        blocos = int((valor / maximo) * tamanho)
        return "█" * blocos + "░" * (tamanho - blocos)

    t_icon = "🔴" if dados["temperatura"] > LIMITES["temperatura_max"] else "🟢"
    e_icon = "🔴" if dados["energia"] < LIMITES["energia_min"] else "🟢"
    s_icon = "🔴" if dados["sinal"] < LIMITES["sinal_min"] else "🟢"
    p_icon = "🔴" if dados["pressao"] < LIMITES["pressao_min"] else "🟢"
    o_icon = "🔴" if dados["oxigenio"] < LIMITES["oxigenio_min"] else "🟢"

    print(f"{t_icon} Temperatura:{dados['temperatura']:>6.1f}°C  [{barra(min(dados['temperatura'], 100))}]")
    print(f"{e_icon} Energia:{dados['energia']:>6.1f}%   [{barra(dados['energia'])}]")
    print(f"{s_icon} Sinal:{dados['sinal']:>6.1f}%   [{barra(dados['sinal'])}]")
    print(f"{p_icon} Pressão:{dados['pressao']:>6.1f} kPa")
    print(f"{o_icon} Oxigênio:{dados['oxigenio']:>6.1f}% [{barra(dados['oxigenio'], 25)}]")
    print(f"Status Módulo: {dados['status_modulo']}")

    print(f"\n{linha2}")
    print(f"  STATUS GERAL: {nivel}")
    print(linha2)

    if alertas:
        print("\n  🚨  ALERTAS DETECTADOS")
        print(linha2)
        for alerta in alertas:
            print(f"  {alerta}")

        print("\n  🛠️   AÇÕES AUTOMATIZADAS")
        print(linha2)
        for acao in acoes:
            print(f"  {acao}")
    else:
        print("\n  ✅  Todos os sistemas operando dentro dos parâmetros normais.")

    print(f"\n{linha2}")
    print("  🤖  ANÁLISE DA IA (Llama)")
    print(linha2)
    import textwrap
    for linha_ia in textwrap.wrap(analise_ia, width=51):
        print(f"  {linha_ia}")

    print(f"\n{linha}\n")

print("Função de exibição pronta.")
```

## ▶️ Célula 7 — EXECUTAR: Cenário Normal

```python
print("Gerando dados para CENÁRIO NORMAL...\n")

dados = gerar_dados_missao(cenario="normal")
alertas, acoes = verificar_alertas(dados)

print("Consultando IA... (pode levar alguns segundos)")
analise_ia = analisar_com_ia(dados, alertas)

exibir_status(dados, alertas, acoes, analise_ia)
```

## 🚨 Célula 8 — EXECUTAR: Cenário Crítico (Situação de Emergência)

```python
print("Simulando CENÁRIO CRÍTICO de emergência...\n")

dados_crit = gerar_dados_missao(cenario="critico")
alertas_crit, acoes_crit = verificar_alertas(dados_crit)

print("Consultando IA para análise de emergência...")
analise_crit = analisar_com_ia(dados_crit, alertas_crit)

exibir_status(dados_crit, alertas_crit, acoes_crit, analise_crit)
```

## 🎲 Célula 9 — EXECUTAR: Monitoramento Contínuo (3 ciclos aleatórios)

```python
print("Iniciando monitoramento contínuo — 3 ciclos com dados aleatórios\n")
print("(Simulando leitura de sensores em tempo real)\n")

for ciclo in range(1, 4):
    print(f"{'='*55}")
    print(f"  CICLO DE MONITORAMENTO #{ciclo}")
    print(f"{'='*55}")

    dados_ciclo = gerar_dados_missao(cenario="aleatorio")
    alertas_ciclo, acoes_ciclo = verificar_alertas(dados_ciclo)

    print(f"Consultando IA para ciclo #{ciclo}...")
    analise_ciclo = analisar_com_ia(dados_ciclo, alertas_ciclo)

    exibir_status(dados_ciclo, alertas_ciclo, acoes_ciclo, analise_ciclo)

    if ciclo < 3:
        print("⏳ Aguardando próximo ciclo de leitura (3s)...\n")
        time.sleep(3)

print("Monitoramento contínuo concluído.")
```

## 💬 Célula 10 (BÔNUS) — Interface de Chat com o Mission Control AI

```python
def chatbot_missao():
    """
    Interface conversacional com o Mission Control AI.
    O operador pode fazer perguntas e receber respostas contextualizadas.
    Digite 'sair' para encerrar.
    """
    dados_sessao = gerar_dados_missao(cenario="aleatorio")
    alertas_sessao, _ = verificar_alertas(dados_sessao)

    historico = [
        {
            "role": "system",
            "content": (
                f"Você é o assistente de IA do Mission Control da missão espacial {NOME_MISSAO}."
                f"Status atual: Temperatura={dados_sessao['temperatura']}°C,"
                f"Energia={dados_sessao['energia']}%,Sinal={dados_sessao['sinal']}%,"
                f"Pressão={dados_sessao['pressao']}kPa,O2={dados_sessao['oxigenio']}%,"
                f"Módulo={dados_sessao['status_modulo']}."
                f"Alertas ativos: {len(alertas_sessao)}."
                "Responda sempre em português. Seja técnico e preciso."
            )
        }
    ]

    print("="*55)
    print(f"MISSION CONTROL AI — CHAT OPERACIONAL")
    print(f"Missão: {NOME_MISSAO} | Alertas: {len(alertas_sessao)}")
    print("Digite'sair'para encerrar a sessão")
    print("="*55 + "\n")

    while True:
        try:
            entrada = input("  🧑‍🚀 Operador: ").strip()
        except EOFError:
            print("  [Sessão encerrada automaticamente]")
            break

        if entrada.lower() in ["sair", "exit", "quit"]:
            print("Sessão de chat encerrada. Boa missão!")
            break

        if not entrada:
            continue

        historico.append({"role": "user", "content": entrada})

        try:
            resposta = ollama.chat(model="llama3.2:1b", messages=historico)
            msg_ia = resposta["message"]["content"]
        except Exception as e:
            msg_ia = f"[ERRO: {str(e)}]"

        historico.append({"role": "assistant", "content": msg_ia})

        print(f"\n  🤖 Mission Control AI:")
        import textwrap
        for l in textwrap.wrap(msg_ia, width=51):
            print(f"  {l}")
        print()

# Executar o chatbot
chatbot_missao()
```
